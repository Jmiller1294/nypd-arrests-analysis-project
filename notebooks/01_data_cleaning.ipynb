```python
import pandas as pd
import geopandas as gpd
```

```python
df = pd.read_excel('../data/raw_nypd_arrests_data.xlsx')
df
```

```python
print(df.shape)
print(df.columns.tolist())
display(df.info()) 
display(df.head())
display(df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
# check uniqueness
df['ARREST_KEY'].is_unique
df.duplicated(subset='ARREST_KEY').sum()
```

```python
df = df.drop(columns=['PD_CD', 'KY_CD', 'LAW_CODE', 'JURISDICTION_CODE', 'X_COORD_CD', 'Y_COORD_CD'])
df
```

```python
df = df.replace({'ARREST_BORO': {'B': 'BRONX', 'K': 'BROOKLYN', 'M': 'MANHATTAN', 'Q': 'QUEENS', 'S': 'STATEN ISLAND'}})
df
```

```python
df = df.replace({'LAW_CAT_CD': {'F': 'FELONY', 'M': 'MISDEMEANOR', 'V': 'VIOLATION','I': 'INFRACTION'}})
df
```

```python
df = df.replace({'PERP_SEX': {'F': 'FEMALE', 'M': 'MALE', 'U': 'UNKNOWN'}})
df
```

```python
df['OFNS_DESC'] = df['OFNS_DESC'].str.replace(r".*assault.*", "ASSAULT", case=False, regex=True)
df
```

```python
df['OFNS_DESC'] = df['OFNS_DESC'].str.replace("AGRICULTURE & MRKTS LAW-UNCLASSIFIED", "ANIMAL CRUELTY")
```

```python
df.fillna('UNKNOWN', inplace=True)
df
```

```python
months_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

df['MONTH'] = df['ARREST_DATE'].dt.month_name()
df
```

```python
df.to_csv("../data/clean_nypd_arrests_data.csv", index=False)
```
